Model Comparison with Regression Lines

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

# Assuming df_model needs to be defined, here's a placeholder definition.
# Replace this with your actual data loading or DataFrame creation.
df_model = pd.DataFrame({
    'release_year': np.random.randint(2000, 2023, 100),
    'engagement': np.random.rand(100) * 100
})

# ---------------- DATA ---------------- #
# single feature use karna zaroori hai line ke liye
X = df_model[['release_year']].values
y = df_model['engagement'].values

# sort data (IMPORTANT for proper line)
sort_idx = np.argsort(X.flatten())
X = X[sort_idx]
y = y[sort_idx]

# smooth X for curve
X_range = np.linspace(X.min(), X.max(), 200).reshape(-1,1)

# ---------------- LINEAR REGRESSION ---------------- #
lr = LinearRegression()
lr.fit(X, y)
y_lr = lr.predict(X_range)

# ---------------- POLYNOMIAL REGRESSION ---------------- #
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)
X_range_poly = poly.transform(X_range)

pr = LinearRegression()
pr.fit(X_poly, y)
y_pr = pr.predict(X_range_poly)

# ---------------- DECISION TREE ---------------- #
dt = DecisionTreeRegressor(max_depth=5)
dt.fit(X, y)
y_dt = dt.predict(X_range)

# ---------------- RANDOM FOREST ---------------- #
rf = RandomForestRegressor(n_estimators=100)
rf.fit(X, y)
y_rf = rf.predict(X_range)

# ---------------- PLOT ALL ---------------- #
plt.figure(figsize=(10,6))

plt.scatter(X, y, s=10, label="Data")

plt.plot(X_range, y_lr, label="Linear Regression", linewidth=2)
plt.plot(X_range, y_pr, label="Polynomial Regression", linewidth=2)
plt.plot(X_range, y_dt, label="Decision Tree", linewidth=2)
plt.plot(X_range, y_rf, label="Random Forest", linewidth=2)

plt.xlabel("Release Year")
plt.ylabel("Engagement")
plt.title("Model Comparison with Regression Lines")
plt.legend()

plt.show()
```

Line Regression

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

X = df_model[['release_year']].values
y = df_model['engagement'].values

# sort
idx = np.argsort(X.flatten())
X, y = X[idx], y[idx]

X_range = np.linspace(X.min(), X.max(), 200).reshape(-1,1)

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X_range)

plt.figure(figsize=(6,4))
plt.scatter(X, y, s=10)
plt.plot(X_range, y_pred, linewidth=2)

plt.title("Linear Regression")
plt.xlabel("Release Year")
plt.ylabel("Engagement")
plt.show()
```

Polynomial Regression

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

X = df_model[['release_year']].values
y = df_model['engagement'].values

idx = np.argsort(X.flatten())
X, y = X[idx], y[idx]

X_range = np.linspace(X.min(), X.max(), 200).reshape(-1,1)

poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)
X_range_poly = poly.transform(X_range)

model = LinearRegression()
model.fit(X_poly, y)

y_pred = model.predict(X_range_poly)

plt.figure(figsize=(6,4))
plt.scatter(X, y, s=10)
plt.plot(X_range, y_pred, linewidth=2)

plt.title("Polynomial Regression (d=2)")
plt.xlabel("Release Year")
plt.ylabel("Engagement")
plt.show()
```
